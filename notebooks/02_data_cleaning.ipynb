```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/amazon_reviews.csv")

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

```python
print(df.columns.tolist())
```

```python
df.head()
```

```python
print("Rating column:")
print(df["Rating"].value_counts(dropna=False))

print("\nReview Rating column:")
print(df["Review Rating"].value_counts(dropna=False))
```

```python
clean_df = df.copy()

print("Working copy created.")
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(clean_df.columns.tolist())
```

```python
duplicate_count = clean_df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
missing_values = clean_df.isnull().sum().sort_values(ascending=False)

print(missing_values)
```

```python
clean_df = clean_df.drop(columns=["manufacturer_response"])

print("manufacturer_response removed.")
print("Columns remaining:", len(clean_df.columns))
```

```python
clean_df["category"] = clean_df["category"].fillna("Unknown")
clean_df["sub_category"] = clean_df["sub_category"].fillna("Unknown")
clean_df["product_description"] = clean_df["product_description"].fillna("Unknown")

print("Missing categorical/text values handled.")
```

```python
print(clean_df.isnull().sum().sort_values(ascending=False))
```

```python
print(clean_df.dtypes)
```

```python
clean_df["review_date"] = pd.to_datetime(
    clean_df["review_date"],
    errors="coerce"
)

print(clean_df["review_date"].dtype)
```

```python
print(clean_df["helpful_review_count"].value_counts(dropna=False))
```

```python
clean_df["helpful_review_count"] = (
    clean_df["helpful_review_count"]
    .replace("One person found this helpful Helpful", "1")
    .astype(int)
)

print(clean_df["helpful_review_count"].value_counts().sort_index())
```

```python
print(clean_df["verified_purchase"].value_counts(dropna=False))
```

```python
clean_df["verified_purchase"] = clean_df["verified_purchase"].replace({
    "Verified Purchase": "Yes",
    "NO": "No"
})

print(clean_df["verified_purchase"].value_counts())
```

```python
def rating_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"

clean_df["rating_sentiment"] = clean_df["review_rating"].apply(rating_sentiment)

print(clean_df["rating_sentiment"].value_counts())
```

```python
print("Shape:", clean_df.shape)
print("\nMissing values:")
print(clean_df.isnull().sum().sum())

print("\nDuplicate rows:")
print(clean_df.duplicated().sum())

print("\nData types:")
print(clean_df.dtypes)
```

```python
output_path = "../data/processed/cleaned_reviews.csv"

clean_df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved to: {output_path}")
```
